# 04 - Trends and events
EU import unit values (validated, 3-month weighted) and the origin mix around dated events (`seeds/events.csv`).

```python
import pandas as pd, duckdb
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
M, C = ROOT / "data" / "marts", ROOT / "data" / "clean"
pd.set_option("display.width", 160); pd.set_option("display.max_columns", 20)
```

```python
m = pd.read_csv(M / 'monthly_prices.csv', parse_dates=['period'])
lvl = m.assign(year=m.period.dt.year).groupby(['material_id', 'year']).apply(lambda g: g.eu_import_value_eur.sum() / g.eu_import_kg.sum(), include_groups=False).unstack().round(2)
lvl
```

```python
o = pd.read_csv(M / 'origin_monthly.csv', parse_dates=['period'])
ga = o[(o.material_id == 'GA') & (o.period >= '2022-07-01') & (o.period <= '2024-06-01')]
ga.assign(cn=ga.origin.eq('CN')).groupby([pd.Grouper(key='period', freq='QS'), 'cn']).quantity_kg.sum().unstack().div(1000).round(1)
```
